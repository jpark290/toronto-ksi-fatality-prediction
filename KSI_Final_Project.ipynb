## 3. Missing Values (Raw Data)
This section examines the amount and percentage of missing values in the original dataset to identify potential data quality issues before preprocessing. This first pass is done on the raw, un-aggregated file (one row per party/person).

```python
def missing_report(frame):
    report = pd.DataFrame({
        "Missing_Count": frame.isnull().sum(),
        "Missing_Pct": (frame.isnull().sum() / len(frame) * 100).round(2),
    })
    return report[report["Missing_Count"] > 0].sort_values("Missing_Pct", ascending=False)

missing_report(df)
```

## 4. Target Variable Preparation
The original ACCLASS column contains multiple accident severity categories. For this project, a binary target variable was created by keeping only Fatal and Non-Fatal Injury records, where Fatal is encoded as 1 and Non-Fatal Injury as 0.

### 4.1 Clean Target Labels

```python
df["ACCLASS_CLEAN"] = (
    df["ACCLASS"]
    .astype(str)
    .str.strip()
)
```

### 4.2 Keep Only KSI Records

```python
df_clean = df[
    df["ACCLASS_CLEAN"].isin(
        ["Fatal", "Non-Fatal Injury"]
    )
].copy()
```

### 4.3 Create Binary Target

```python
df_clean["TARGET_FATAL"] = (
    df_clean["ACCLASS_CLEAN"] == "Fatal"
).astype(int)

df_clean["CLASS_LABEL"] = (
    df_clean["TARGET_FATAL"]
    .map({
        1: "Fatal",
        0: "Non-Fatal"
    })
)
```

## 5. Handling Missing ACCNUM Before Aggregation

`ACCNUM` is the key that Section 6 groups on to collapse party-level rows into one row per collision. Before aggregating, `df_clean['ACCNUM']` is checked for missing values, since any row with a missing `ACCNUM` cannot be assigned to a group and would otherwise be silently dropped by `groupby("ACCNUM")`.

A quick bias check (fatal rate and year distribution for missing vs. non-missing `ACCNUM` rows) is run first, followed by a check for whether the missing-`ACCNUM` rows share other collision-identifying fields (`DATE`, `TIME`, `STREET1`, `STREET2`) with one another - if they do, they're likely multiple people from the *same* real collision that lost its case number, and should be recovered together rather than treated as separate incidents.

```python
# --- Diagnostic: quantify missing ACCNUM and check for bias before deciding how to handle it ---
missing_mask = df_clean['ACCNUM'].isna()
n_missing = missing_mask.sum()
pct_missing = n_missing / len(df_clean) * 100
print(f"{n_missing} of {len(df_clean)} rows ({pct_missing:.1f}%) have missing ACCNUM in df_clean")

print("\nFatal rate by ACCNUM missing status (False = has ACCNUM, True = missing):")
print(df_clean.groupby(missing_mask)['TARGET_FATAL'].mean().round(4))

df_clean['_YEAR_TEMP'] = pd.to_datetime(df_clean['DATE']).dt.year
print("\nYear distribution by ACCNUM missing status:")
display(df_clean.groupby(missing_mask)['_YEAR_TEMP'].describe())

# Check whether missing-ACCNUM rows share DATE/TIME/STREET1/STREET2 with each other,
# which would indicate they're multiple people from the SAME real collision
key_cols = ['DATE', 'TIME', 'STREET1', 'STREET2']
group_sizes = (
    df_clean.loc[missing_mask, key_cols]
    .fillna('UNKNOWN')
    .groupby(key_cols)
    .size()
)
n_multi_row_groups = (group_sizes > 1).sum()
n_rows_in_multi = group_sizes[group_sizes > 1].sum()
n_isolated_rows = (group_sizes == 1).sum()

print(f"\nOf {n_missing} missing-ACCNUM rows:")
print(f"  - {n_rows_in_multi} rows across {n_multi_row_groups} groups share DATE/TIME/STREET1/STREET2 "
      f"with at least one other row (likely the same real collision, split across multiple person-rows)")
print(f"  - {n_isolated_rows} rows have no match on those fields (likely isolated single-party records)")
```

```python
# --- Recover collision IDs for missing-ACCNUM rows ---
# Rows sharing DATE/TIME/STREET1/STREET2 are grouped back together (same recovered collision);
# rows with no match naturally end up alone in their own group (equivalent to a synthetic ID),
# so a single groupby+ngroup handles both cases without a separate fallback step.
recovery_key = df_clean.loc[missing_mask, key_cols].fillna('UNKNOWN')

recovered_ids = (
    recovery_key
    .groupby(key_cols)
    .ngroup()
    .astype(str)
    .radd('recovered_')
)

# Allow both original numeric IDs and recovered string IDs
df_clean["ACCNUM"] = df_clean["ACCNUM"].astype("object")

df_clean.loc[missing_mask, 'ACCNUM'] = recovered_ids.values

# Sanity check: no missing ACCNUM should remain
assert df_clean['ACCNUM'].isna().sum() == 0, "All rows should now have a non-null ACCNUM"

n_recovered_groups = df_clean.loc[missing_mask, 'ACCNUM'].nunique()
print(f"Recovered {n_missing} previously-missing rows into {n_recovered_groups} collision IDs "
      f"({n_missing - n_recovered_groups} rows were merged into shared groups; "
      f"the rest became single-row collisions since no match was found).")
print(f"Remaining missing ACCNUM after recovery: {df_clean['ACCNUM'].isna().sum()}")

df_clean.drop(
    columns=["_YEAR_TEMP"],
    inplace=True,
    errors="ignore"
)
```

**Decision:** rather than dropping the ~26% of rows with missing `ACCNUM` (which would risk biasing the collision-level dataset - see the checks above), collision IDs are reconstructed from `DATE`, `TIME`, `STREET1`, and `STREET2`, which should match across all person-rows of the same real collision the same way `ACCNUM` does. Rows that don't match anything else on these fields are kept as their own single-party collision rather than being discarded. `df_clean['ACCNUM']` now has no missing values, so the Section 6 aggregation below will retain every row instead of silently dropping ~26% of the data.

## 6. Aggregating Records to the Collision Level
Each row in df_clean is a party/person in a collision, so the same *ACCNUM* repeats. Since the modelling question is about the collision itself (not the individual person), records are aggregated to one row per *ACCNUM*:

- **Collision-level attributes** (date, location, road/weather/lighting conditions, contributing factors) are taken with .first() - they're identical across all party rows of the same collision.

- **Party/vehicle-type involvement flags** (pedestrian, cyclist, automobile, etc) are combined with .max() so that a collision is flagged "Yes" if any party matches that type.

- The **target** is aggregated with .max(): a collision is Fatal (1) if at least one person in it died.

```python
# Aggregate collision features
collision_features = [
    "DATE",
    "TIME",
    "ROAD_CLASS",
    "DISTRICT",
    "LATITUDE",
    "LONGITUDE",
    "ACCLOC",
    "TRAFFCTL",
    "VISIBILITY",
    "LIGHT",
    "RDSFCOND",
    "INITDIR",
    "DIVISION",
    "HOOD_158",
    "NEIGHBOURHOOD_158",
    "SPEEDING",
    "AG_DRIV",
    "REDLIGHT",
    "ALCOHOL",
    "DISABILITY"
]

party_features = [
    "PEDESTRIAN",
    "CYCLIST",
    "AUTOMOBILE",
    "MOTORCYCLE",
    "TRUCK",
    "TRSN_CITY_VEH",
    "PASSENGER"
]

# Aggregate by collision
collision_target = (
    df_clean
    .groupby("ACCNUM")["TARGET_FATAL"]
    .max()
)

collision_df = (
    df_clean
    .groupby("ACCNUM")[collision_features]
    .first()
)

party_df = (
    df_clean
    .groupby("ACCNUM")[party_features]
    .max()
)

collision_df = pd.concat(
    [
        collision_df,
        party_df,
        collision_target
    ],
    axis=1
)
# Keep HOOD_158 as a categorical neighbourhood identifier. Converting it to
# numeric would impose a false order and would turn the valid 'NSA' code into NaN.
collision_df["HOOD_158"] = collision_df["HOOD_158"].astype("string")
print(f"Collision-level shape: {collision_df.shape}")
print("\nClass balance at the collision level:")
print(collision_df["TARGET_FATAL"].value_counts(normalize=True).round(4))
```

## 7. Missing Values After Aggregation and Feature Cleaning
Missing values are re-checked at the collision level (the proportions can shift after aggregation), and each group of columns is imputed according to what "missing" actually means for that column:

- **Contributing-factor / road-user flags** (SPEEDING, ALCOHOL, CYCLIST, ...): missing means the factor or road-user type was not present, so these are filled with "No" rather than treated as unknown data.

- **Environmental/road condition fields** (ROAD_CLASS, DISTRICT, TRAFFCTL, VISIBILITY, RDSFCOND, LIGHT): these are filled with the column mode, since they represent the most plausible/typical condition and missingness here looks like reporting gaps rather than a meaningful category.

- **INITDIR, ACCLOC, DIVISION**: filled with an explicit "Unknown" category — there isn't a safe default to assume, so this preserves the fact that the value is genuinely unrecorded rather than guessing at it.

```python
display(missing_report(collision_df))
```

```python
# Missing values are left as-is here rather than filled manually. The columns are
# grouped by what missingness means for that column (absence of a factor, a
# reasonable mode fill, or a genuinely-unknown category), and that grouping is
# passed straight into the imputers of the ColumnTransformer pipeline in
# Section 17, so imputation happens once, inside the pipeline, on train/test
# data only (avoiding any double-imputation or leakage from filling on the
# full collision_df here).
print("Missing values before pipeline imputation:")
remaining = missing_report(collision_df)
print("None" if remaining.empty else remaining)
```

## 8. Feature Engineering (Date and Time)
DATE and TIME are decomposed into YEAR, MONTH, DAY_OF_WEEK, and HOUR, which are more useful as individual numeric/cyclical predictors than the raw date/time fields, and the originals are dropped. Whitespace is also stripped from all text columns as a final cleanup pass.

```python
collision_df["DATE"] = pd.to_datetime(collision_df["DATE"])
collision_df["YEAR"] = collision_df["DATE"].dt.year # maybe should be dropped
collision_df["MONTH"] = collision_df["DATE"].dt.month # weather factor
collision_df["DAY_OF_WEEK"] = collision_df["DATE"].dt.dayofweek # weekend driving patterns
collision_df["HOUR"] = collision_df["TIME"] // 100 # rush hours
collision_df.drop(columns=["DATE", "TIME"], inplace=True)

for col in collision_df.select_dtypes(include="object").columns:
    collision_df[col] = collision_df[col].str.strip()

collision_df.head()
```

## 13. Feature Selection & Justification

Before splitting or modelling, every column is either **kept** or **discarded**, with a reason:

**Discarded:**

| Column | Reason for discarding |
|---|---|
| `ACCNUM` | Row identifier, not predictive; already used only to aggregate/index the data. |
| `ACCLASS`, `ACCLASS_CLEAN`, `CLASS_LABEL` | These *are* the target (or were used only to derive it) — keeping them would leak the answer directly into the features. |
| `HOOD_158`, `NEIGHBOURHOOD_158` | Both identify the 158-neighbourhood area. Location is already represented by exact `LATITUDE`/`LONGITUDE` and broader `DISTRICT`/`DIVISION` categories. Keeping another 159-category location representation would add substantial dimensionality and overlapping information. |
| `DATE`, `TIME` | Replaced by the engineered `YEAR`, `MONTH`, `DAY_OF_WEEK`, `HOUR` features (Section 8), which are more directly usable by a model. |

**Kept**, grouped by rationale:

| Group | Columns | Why kept |
|---|---|---|
| Location | `LATITUDE`, `LONGITUDE`, `DISTRICT`, `DIVISION` | `LATITUDE` and `LONGITUDE` retain the exact collision location, while `DISTRICT` and `DIVISION` provide broader area categories. This keeps location information without adding the 159-category neighbourhood feature. |
| Road / environment | `ROAD_CLASS`, `ACCLOC`, `TRAFFCTL`, `VISIBILITY`, `LIGHT`, `RDSFCOND`, `INITDIR` | Directly describe the physical conditions of the collision, which is exactly the kind of situational information a severity model should use. |
| Time | `YEAR`, `MONTH`, `DAY_OF_WEEK`, `HOUR` | Captures traffic-volume and lighting-related patterns (Section 10) without leaking outcome information. |
| Contributing factors | `SPEEDING`, `AG_DRIV`, `REDLIGHT`, `ALCOHOL`, `DISABILITY` | Section 11 shows fatality-rate differences across these recorded factors, so they may provide predictive information about collision severity. |
| Road-user involvement | `PEDESTRIAN`, `CYCLIST`, `AUTOMOBILE`, `MOTORCYCLE`, `TRUCK`, `TRSN_CITY_VEH`, `PASSENGER` | These columns record which road-user types were involved and allow the models to learn their associations with collision severity. |

This keeps every feature that is plausibly associated with collision outcomes or empirically showed a
meaningful fatality-rate difference in the EDA, while removing anything that is an identifier, a duplicate, or
a stand-in for the target itself.

## 15. Handling Imbalanced Classes

The EDA (Section 6) showed the target is imbalanced (fatal collisions are the minority class). Following the
strategies outlined for imbalanced classification (see
[elitedatascience.com/imbalanced-classes](https://elitedatascience.com/imbalanced-classes)), the main options
are:

1. **Up-sample the minority class** (e.g. random oversampling, or synthetic oversampling with **SMOTE**).
2. **Down-sample the majority class** (randomly drop majority-class rows - simple, but throws away data).
3. **Use class weights** (e.g. `class_weight="balanced"`) so the model penalizes minority-class errors more,
   without changing the data itself.
4. **Use algorithms that tend to be more robust to imbalance** (tree ensembles such as Random Forest / Gradient
   Boosting), and evaluate with metrics that aren't misleading under imbalance (precision, recall, F1, ROC-AUC -
   **not** plain accuracy, since predicting "Non-Fatal" for everyone would already score about 86.1% accuracy here).

This project uses **SMOTE (Synthetic Minority Over-sampling Technique)**, applied **only to the training
fold** (never to the test set, to avoid leaking synthetic information into evaluation) and **only after**
pre-processing/encoding, so it operates on numeric feature space. It is combined inside the pipeline with
imbalance-aware evaluation metrics in Parts 3 and 4.

## 16. Pre-processing Pipeline (`ColumnTransformer`)

A single `ColumnTransformer` handles the feature groups so the exact same, leakage-safe transformations are
applied consistently to train and test data:

- **Numeric features:** missing values imputed with the **median**, then **standardized** (`StandardScaler`)
  so that features on different scales (e.g. `YEAR` vs. `LATITUDE`) contribute comparably to
  distance-/gradient-based models.
- **Binary indicator features:** missing values are treated as `No`, then one-hot encoded.
- **Mode-imputed categorical features:** missing values are replaced with the most frequent training category, then one-hot encoded.
- **Unknown-category features:** missing values are retained as an explicit `Unknown` category, then one-hot encoded.

All encoders use `handle_unknown="ignore"` so that a category seen only in test or API data does not break the pipeline.

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# Columns with missing meaning = absence
binary_features = [
    "DISABILITY", "ALCOHOL", "TRSN_CITY_VEH", "TRUCK", "REDLIGHT",
    "MOTORCYCLE", "SPEEDING", "CYCLIST", "PASSENGER",
    "PEDESTRIAN", "AG_DRIV", "AUTOMOBILE"
]

# Columns where mode is reasonable
mode_features = [
    "ROAD_CLASS", "DISTRICT", "TRAFFCTL",
    "VISIBILITY", "RDSFCOND", "LIGHT"
]

# Columns where missingness itself is meaningful
unknown_features = [
    "INITDIR", "ACCLOC", "DIVISION"
]

numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

binary_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="constant", fill_value="No")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

mode_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

unknown_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="constant", fill_value="Unknown")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer(transformers=[
    ("num", numeric_transformer, numeric_features),
    ("binary", binary_transformer, binary_features),
    ("mode", mode_transformer, mode_features),
    ("unknown", unknown_transformer, unknown_features)
])

preprocessor
```

## 17. Full Modelling Pipeline

The `ColumnTransformer` above is chained together with **SMOTE** (fit only on the training folds during
cross-validation/fitting, via `imblearn`'s `Pipeline`) and a classifier, so that the entire
transform → resample → predict sequence is a single object. This is what "streamlines the pre-processing" —
calling `.fit()` once runs every step in the right order, and `.predict()` on new/raw data automatically
applies the same imputation, scaling, and encoding used during training (with **no** resampling at
prediction time, since `imblearn.Pipeline` only resamples during `.fit`).

A Logistic Regression classifier is used here as a simple, interpretable baseline; swapping in a different
estimator (e.g. `RandomForestClassifier`) only requires changing the final pipeline step.

```python
from imblearn.pipeline import Pipeline as ImbPipeline
from sklearn.linear_model import LogisticRegression

full_pipeline = ImbPipeline(steps=[
    ("preprocessing", preprocessor),
    ("smote", SMOTE(random_state=42)),
    ("classifier", LogisticRegression(max_iter=1000, random_state=42)),
])

full_pipeline.fit(X_train, y_train)
full_pipeline
```

# Part 3: Predictive Model Building

## 19. Modelling Requirements and Tuning Strategy

This section implements the five classification algorithms required by the project rubric:

1. Logistic Regression
2. Decision Tree
3. Support Vector Machine
4. Random Forest
5. Neural Network

Every model uses the preprocessing and SMOTE pipeline prepared in Part 2. Hyperparameters are selected using only the training data. `GridSearchCV` is used for models with smaller and clearly defined search spaces. `RandomizedSearchCV` is used for models with larger search spaces, where testing every possible combination would be inefficient.

Fatal-class F1 is used as the tuning score because the target is imbalanced and F1 balances fatal precision and fatal recall. The test set was used once for a preliminary pipeline sanity check, but it was not used for hyperparameter tuning. It is used to compare the tuned models in Part 4.

```python
from scipy.stats import randint, loguniform

from sklearn.base import clone
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import (
    GridSearchCV,
    RandomizedSearchCV,
    StratifiedKFold,
)
from sklearn.neural_network import MLPClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier

RANDOM_STATE = 42

# The same folds are used for every model so their tuning results are comparable.
model_cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=RANDOM_STATE,
)


def build_model_pipeline(classifier):
    # Create an unfitted preprocessing, SMOTE, and classifier pipeline.
    return ImbPipeline(steps=[
        ("preprocessing", clone(preprocessor)),
        ("smote", SMOTE(random_state=RANDOM_STATE)),
        ("classifier", classifier),
    ])


tuning_results = []
tuned_models = {}


def record_tuning_result(model_name, search_name, search_object):
    # Store the best training cross-validation result for later review.
    tuned_models[model_name] = search_object.best_estimator_
    tuning_results.append({
        "Model": model_name,
        "Search Method": search_name,
        "Best CV Fatal F1": search_object.best_score_,
        "Best Parameters": search_object.best_params_,
    })
    print(f"{model_name} completed")
    print(f"Best CV fatal F1: {search_object.best_score_:.4f}")
    print(f"Best parameters: {search_object.best_params_}")
```

## 20. Logistic Regression with Grid Search

Logistic Regression provides an interpretable linear baseline. Grid Search tests different regularization strengths and compatible solvers. Regularization controls coefficient size and helps reduce overfitting.

```python
logistic_pipeline = build_model_pipeline(
    LogisticRegression(max_iter=2000, random_state=RANDOM_STATE)
)

logistic_param_grid = {
    "classifier__C": [0.01, 0.1, 1.0, 10.0],
    "classifier__solver": ["liblinear", "lbfgs"],
    "classifier__penalty": ["l2"],
}

logistic_search = GridSearchCV(
    estimator=logistic_pipeline,
    param_grid=logistic_param_grid,
    scoring="f1",
    cv=model_cv,
    n_jobs=-1,
    refit=True,
    return_train_score=False,
)
logistic_search.fit(X_train, y_train)

record_tuning_result(
    "Logistic Regression",
    "GridSearchCV",
    logistic_search,
)
```

## 21. Decision Tree with Grid Search

A Decision Tree can model non-linear thresholds and interactions. Grid Search controls tree complexity through maximum depth, split size, and leaf size. These parameters are important because an unrestricted tree can overfit the training data.

```python
decision_tree_pipeline = build_model_pipeline(
    DecisionTreeClassifier(random_state=RANDOM_STATE)
)

decision_tree_param_grid = {
    "classifier__criterion": ["gini", "entropy"],
    "classifier__max_depth": [5, 10, 20, None],
    "classifier__min_samples_split": [2, 10],
    "classifier__min_samples_leaf": [1, 5],
}

decision_tree_search = GridSearchCV(
    estimator=decision_tree_pipeline,
    param_grid=decision_tree_param_grid,
    scoring="f1",
    cv=model_cv,
    n_jobs=-1,
    refit=True,
    return_train_score=False,
)
decision_tree_search.fit(X_train, y_train)

record_tuning_result(
    "Decision Tree",
    "GridSearchCV",
    decision_tree_search,
)
```

## 22. Support Vector Machine with Grid Search

Support Vector Machine models are tested with linear, radial basis function, and polynomial kernels. The linear kernel tests a linear boundary, while the other kernels can represent non-linear boundaries. `C` controls the penalty for classification errors. `gamma` controls the influence of individual training points for non-linear kernels.

```python
svm_pipeline = build_model_pipeline(
    SVC(random_state=RANDOM_STATE)
)

# Separate dictionaries prevent irrelevant parameter combinations.
svm_param_grid = [
    {
        "classifier__kernel": ["linear"],
        "classifier__C": [0.1, 1.0, 10.0],
    },
    {
        "classifier__kernel": ["rbf"],
        "classifier__C": [0.1, 1.0, 10.0],
        "classifier__gamma": ["scale", 0.01, 0.1],
    },
    {
        "classifier__kernel": ["poly"],
        "classifier__C": [0.1, 1.0],
        "classifier__degree": [2, 3],
        "classifier__gamma": ["scale"],
    },
]

svm_search = GridSearchCV(
    estimator=svm_pipeline,
    param_grid=svm_param_grid,
    scoring="f1",
    cv=model_cv,
    n_jobs=-1,
    refit=True,
    return_train_score=False,
)
svm_search.fit(X_train, y_train)

record_tuning_result(
    "Support Vector Machine",
    "GridSearchCV",
    svm_search,
)
```

## 23. Random Forest with Randomized Search

Random Forest combines many Decision Trees to reduce the instability of a single tree. Its search space contains many possible combinations, so Randomized Search samples a fixed number of combinations efficiently. The search tunes the number of trees, tree depth, split size, leaf size, and the number of features considered at each split.

```python
random_forest_pipeline = build_model_pipeline(
    RandomForestClassifier(
        random_state=RANDOM_STATE,
        n_jobs=-1,
    )
)

random_forest_param_distributions = {
    "classifier__n_estimators": randint(150, 501),
    "classifier__max_depth": [None, 10, 20, 30],
    "classifier__min_samples_split": randint(2, 16),
    "classifier__min_samples_leaf": randint(1, 9),
    "classifier__max_features": ["sqrt", "log2", 0.5],
}

random_forest_search = RandomizedSearchCV(
    estimator=random_forest_pipeline,
    param_distributions=random_forest_param_distributions,
    n_iter=15,
    scoring="f1",
    cv=model_cv,
    random_state=RANDOM_STATE,
    n_jobs=-1,
    refit=True,
    return_train_score=False,
)
random_forest_search.fit(X_train, y_train)

record_tuning_result(
    "Random Forest",
    "RandomizedSearchCV",
    random_forest_search,
)
```

## 24. Neural Network with Randomized Search

The scikit-learn `MLPClassifier` is used as the required neural network. Randomized Search is used because hidden-layer structure, activation function, regularization, learning rate, and batch size create a large number of possible combinations. Early stopping keeps a validation portion of each training fold and stops training when performance no longer improves.

```python
neural_network_pipeline = build_model_pipeline(
    MLPClassifier(
        max_iter=500,
        early_stopping=True,
        validation_fraction=0.1,
        n_iter_no_change=15,
        random_state=RANDOM_STATE,
    )
)

neural_network_param_distributions = {
    "classifier__hidden_layer_sizes": [
        (32,),
        (64,),
        (64, 32),
        (128, 64),
    ],
    "classifier__activation": ["relu", "tanh"],
    "classifier__alpha": loguniform(1e-5, 1e-2),
    "classifier__learning_rate_init": loguniform(1e-4, 1e-2),
    "classifier__batch_size": [32, 64, 128],
}

neural_network_search = RandomizedSearchCV(
    estimator=neural_network_pipeline,
    param_distributions=neural_network_param_distributions,
    n_iter=12,
    scoring="f1",
    cv=model_cv,
    random_state=RANDOM_STATE,
    n_jobs=-1,
    refit=True,
    return_train_score=False,
)
neural_network_search.fit(X_train, y_train)

record_tuning_result(
    "Neural Network",
    "RandomizedSearchCV",
    neural_network_search,
)
```

# Part 4: Model Scoring and Evaluation

## 26. Evaluation Approach

The five tuned models are evaluated on the same held-out test set. Fatal collisions are treated as the positive class. Accuracy, precision, recall, F1, confusion matrices, and ROC curves are reported as required by the rubric.

Fatal-class F1 is the primary selection measure because it balances the ability to detect fatal collisions with the reliability of fatal predictions. Fatal recall is also important because a false negative means that a fatal collision was missed. Accuracy is reported but is not used by itself because 86.1% of the collision records are non-fatal. The test set was previously used once for a preliminary pipeline sanity check, but it was not used during hyperparameter tuning.

```python
from sklearn.metrics import (
    ConfusionMatrixDisplay,
    accuracy_score,
    classification_report,
    f1_score,
    precision_score,
    recall_score,
    roc_auc_score,
    roc_curve,
)

test_predictions = {}
test_scores = {}
evaluation_rows = []
cv_f1_by_model = (
    tuning_summary.set_index("Model")["Best CV Fatal F1"].to_dict()
)

for model_name, model in tuned_models.items():
    predictions = model.predict(X_test)
    if hasattr(model, "predict_proba"):
        scores = model.predict_proba(X_test)[:, 1]
    else:
        scores = model.decision_function(X_test)

    report = classification_report(
        y_test, predictions, output_dict=True, zero_division=0
    )
    test_predictions[model_name] = predictions
    test_scores[model_name] = scores
    evaluation_rows.append({
        "Model": model_name,
        "CV Fatal F1": cv_f1_by_model[model_name],
        "Accuracy": accuracy_score(y_test, predictions),
        "Fatal Precision": precision_score(y_test, predictions, zero_division=0),
        "Fatal Recall": recall_score(y_test, predictions, zero_division=0),
        "Fatal F1": f1_score(y_test, predictions, zero_division=0),
        "Macro F1": report["macro avg"]["f1-score"],
        "Weighted F1": report["weighted avg"]["f1-score"],
        "ROC-AUC": roc_auc_score(y_test, scores),
    })

evaluation_results = (
    pd.DataFrame(evaluation_rows)
    .sort_values(
        ["Fatal F1", "Fatal Recall", "ROC-AUC"],
        ascending=False,
    )
    .reset_index(drop=True)
)

display(evaluation_results.round(4))
```

## 27. Confusion Matrices

Each confusion matrix shows correct and incorrect predictions for both classes. The bottom-right cell is the number of fatal collisions correctly detected. The bottom-left cell is the number of fatal collisions missed by the model.

```python
fig, axes = plt.subplots(2, 3, figsize=(15, 9))

for ax, (model_name, predictions) in zip(axes.flat, test_predictions.items()):
    ConfusionMatrixDisplay.from_predictions(
        y_test,
        predictions,
        display_labels=["Non-Fatal", "Fatal"],
        cmap="Blues",
        colorbar=False,
        ax=ax,
    )
    ax.set_title(model_name)

axes.flat[-1].axis("off")
fig.suptitle("Confusion Matrices on the Test Set", fontsize=16)
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```

## 28. ROC Curves

ROC curves compare the true positive rate and false positive rate across possible decision thresholds. ROC-AUC summarizes this ranking ability. A value of 0.5 is equivalent to random ranking, while a value closer to 1.0 is better.

```python
plt.figure(figsize=(9, 7))

for model_name, scores in test_scores.items():
    fpr, tpr, _ = roc_curve(y_test, scores)
    auc_value = roc_auc_score(y_test, scores)
    plt.plot(
        fpr,
        tpr,
        linewidth=2,
        label=f"{model_name} (AUC = {auc_value:.3f})",
    )

plt.plot([0, 1], [0, 1], "k--", label="Random classifier")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curves on the Test Set")
plt.legend(loc="lower right")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
